```python
import csv
from collections import defaultdict
from pathlib import Path
from statistics import mean

data = Path("../data")
readings = defaultdict(list)
with (data / "raw/readings.csv").open() as stream:
    for row in csv.DictReader(stream):
        readings[row["station"]].append(float(row["reading_m"]))
with (data / "output/station-means.csv").open("w", newline="") as stream:
    writer = csv.writer(stream, lineterminator="\n")
    writer.writerow(["station", "mean_m"])
    writer.writerows((station, mean(values)) for station, values in readings.items())
```
